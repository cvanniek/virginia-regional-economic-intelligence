````python
# Audit imports across the five analytical notebooks.
# ---------------------------------------------------
# This helps us build an accurate requirements.txt
# based on packages the project actually uses.

import json
import re
from pathlib import Path

NOTEBOOK_DIR = PROJECT_ROOT / "notebooks"

notebook_files = [
    NOTEBOOK_DIR / "01_data_acquisition.ipynb",
    NOTEBOOK_DIR / "02_data_quality.ipynb",
    NOTEBOOK_DIR / "03_regional_economic_analysis.ipynb",
    NOTEBOOK_DIR / "04_document_intelligence.ipynb",
    NOTEBOOK_DIR / "05_regional_economic_opportunity_framework.ipynb",
]

imports_by_notebook = {}

for notebook_path in notebook_files:
    notebook = json.loads(
        notebook_path.read_text(encoding="utf-8")
    )

    imports = set()

    for cell in notebook["cells"]:
        if cell.get("cell_type") != "code":
            continue

        source = "".join(cell.get("source", []))

        for line in source.splitlines():
            line = line.strip()

            # Match: import package
            match_import = re.match(
                r"^import\s+([A-Za-z0-9_\.]+)",
                line
            )

            # Match: from package import ...
            match_from = re.match(
                r"^from\s+([A-Za-z0-9_\.]+)\s+import",
                line
            )

            if match_import:
                imports.add(
                    match_import.group(1).split(".")[0]
                )

            if match_from:
                imports.add(
                    match_from.group(1).split(".")[0]
                )

    imports_by_notebook[notebook_path.name] = sorted(imports)

for notebook_name, imports in imports_by_notebook.items():
    print(f"\n{notebook_name}")
    print("-" * len(notebook_name))

    for package in imports:
        print(package)
````

````python
limitations_section = '''

## Limitations

Several limitations should be considered when interpreting the results:

- **QCEW suppression:** some locality-industry observations are unavailable because of confidentiality protections, which can affect measures of diversification and specialization.
- **Locality scale:** counties and independent cities vary substantially in employment base and economic structure; minimum employment thresholds are therefore used for comparative ranking.
- **Broad industry aggregation:** the analysis uses broad NAICS sectors, which improves comparability but can mask important differences within industries.
- **VEDP announcements are forward-looking signals:** announced investment and jobs may be delayed, modified, or not fully realized.
- **Composite weights involve judgment:** the final opportunity score uses transparent baseline weights rather than statistically estimated causal weights. Sensitivity analysis is included to test the robustness of rankings.
- **Localities are administrative geographies:** economic activity frequently crosses county and city boundaries, so the framework should not be interpreted as a substitute for metropolitan, commuting-zone, or input-output analysis.
- **The framework is descriptive, not causal:** observed relationships identify regional patterns and signals but do not establish that any individual policy, investment, or industry caused subsequent economic performance.

## Reproducibility

The notebooks are designed to be run sequentially from `01` through `05`.

Install the required Python packages with:

```bash
pip install -r requirements.txt
```

The workflow then proceeds through:

```text
01_data_acquisition
        ↓
02_data_quality
        ↓
03_regional_economic_analysis
        ↓
04_document_intelligence
        ↓
05_regional_economic_opportunity_framework
```

Intermediate datasets are written to `data/processed/`, while analytical tables and figures are saved under `outputs/`.

Because Notebook 04 retrieves public VEDP web content, results from a future rerun may differ if source pages are edited, removed, or new announcements are published.
'''

with readme_path.open(
    "a",
    encoding="utf-8"
) as f:
    f.write(limitations_section)

print(readme_path.read_text(encoding="utf-8"))
````

````python
# Audit notebook execution health.
# --------------------------------

import json

for notebook_path in notebook_files:
    notebook = json.loads(
        notebook_path.read_text(encoding="utf-8")
    )

    code_cells = [
        cell for cell in notebook["cells"]
        if cell.get("cell_type") == "code"
    ]

    markdown_cells = [
        cell for cell in notebook["cells"]
        if cell.get("cell_type") == "markdown"
    ]

    error_outputs = []
    unexecuted_cells = []

    for i, cell in enumerate(code_cells, start=1):
        execution_count = cell.get("execution_count")

        if execution_count is None and "".join(cell.get("source", [])).strip():
            unexecuted_cells.append(i)

        for output in cell.get("outputs", []):
            if output.get("output_type") == "error":
                error_outputs.append(
                    {
                        "code_cell": i,
                        "ename": output.get("ename"),
                        "evalue": output.get("evalue")
                    }
                )

    print(f"\n{notebook_path.name}")
    print("-" * len(notebook_path.name))
    print(f"Markdown cells: {len(markdown_cells)}")
    print(f"Code cells:     {len(code_cells)}")
    print(f"Stored errors:  {len(error_outputs)}")
    print(f"Unexecuted:     {len(unexecuted_cells)}")

    if error_outputs:
        print("\nErrors:")
        for error in error_outputs:
            print(
                f"  Cell {error['code_cell']}: "
                f"{error['ename']} — {error['evalue']}"
            )

    if unexecuted_cells:
        print("\nUnexecuted code cells:")
        print(unexecuted_cells)
````

````python
# Check the five analytical notebooks for empty or duplicate cells.
# ---------------------------------------------------------------

from collections import Counter

for notebook_path in notebook_files:
    notebook = json.loads(
        notebook_path.read_text(encoding="utf-8")
    )

    empty_cells = []
    cell_contents = []

    for cell_index, cell in enumerate(notebook["cells"], start=1):
        source = "".join(cell.get("source", [])).strip()

        if not source:
            empty_cells.append(cell_index)
        else:
            cell_contents.append(
                (
                    cell_index,
                    cell.get("cell_type"),
                    source
                )
            )

    # Count exact duplicate non-empty cells by type + content.
    content_counter = Counter(
        (cell_type, source)
        for _, cell_type, source in cell_contents
    )

    duplicate_groups = [
        (cell_type, source, count)
        for (cell_type, source), count in content_counter.items()
        if count > 1
    ]

    print(f"\n{notebook_path.name}")
    print("-" * len(notebook_path.name))
    print(f"Empty cells:            {len(empty_cells)}")
    print(f"Exact duplicate groups: {len(duplicate_groups)}")

    if empty_cells:
        print("Empty cell numbers:")
        print(empty_cells)

    if duplicate_groups:
        print("\nDuplicate cells:")

        for cell_type, source, count in duplicate_groups:
            matching_indices = [
                index
                for index, current_type, current_source in cell_contents
                if current_type == cell_type
                and current_source == source
            ]

            preview = source.replace("\n", " ")[:100]

            print(
                f"  {cell_type} cells {matching_indices} "
                f"({count} copies): {preview}"
            )
````
